```python
# ------------------------------------------------------------
# Import LangGraph components
# ------------------------------------------------------------

# START, END → special markers for beginning and end of the graph
# StateGraph → used to build the workflow graph
from langgraph.graph import START, END, StateGraph

# TypedDict → lets us define a dictionary with a fixed structure
from typing import TypedDict


# ------------------------------------------------------------
# Define the state structure for this workflow
# ------------------------------------------------------------

# SpeedState is a dictionary that holds:
# - distance_km: input from user
# - time_hours: input from user
# - speed: output calculated by the node
class SpeedState(TypedDict):
    distance_km: float
    time_hours: float
    speed: float


# ------------------------------------------------------------
# Define the node (function) that performs the calculation
# ------------------------------------------------------------

# This node receives the state (distance + time),
# calculates speed, and returns a new state containing the result.
def Calculate_Speed(state: SpeedState) -> SpeedState:
    distance = state['distance_km']      # extract distance
    time = state['time_hours']           # extract time
    return {'speed': distance / time}    # return only the speed field


# ------------------------------------------------------------
# Build the LangGraph workflow
# ------------------------------------------------------------

# Create a graph that uses SpeedState as its state structure
graph = StateGraph(SpeedState)

# Add a node named "Calculate_Speed" that runs the function above
graph.add_node('Calculate_Speed', Calculate_Speed)

# Connect START → Calculate_Speed → END
graph.add_edge(START, 'Calculate_Speed')
graph.add_edge('Calculate_Speed', END)

# Compile the graph into a runnable workflow
workflow = graph.compile()

# Display the compiled graph object (Jupyter shows a diagram)
workflow
```
